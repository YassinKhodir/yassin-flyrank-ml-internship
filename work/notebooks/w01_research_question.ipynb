## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

The starter data contains enough pages and variation to make prioritization useful. The checks below report the exact scale, typical search visibility, and share of pages currently marked as trending down. That current-window trend is descriptive evidence for choosing the lane, not a causal claim or an ideal future-outcome label.

```python
from pathlib import Path
import pandas as pd

data_candidates = [
    Path("data/raw/content_refresh_anonymized.csv"),
    Path("../../data/raw/content_refresh_anonymized.csv"),
]
data_path = next((path for path in data_candidates if path.exists()), None)
if data_path is None:
    raise FileNotFoundError("Run this notebook from the repository root or work/notebooks directory.")

df = pd.read_csv(data_path)
declining_pages = int(df["trend_direction"].eq("down").sum())
declining_share = df["trend_direction"].eq("down").mean()

print(f"Pages: {len(df):,} across {df['client_id'].nunique()} pseudonymized clients")
print(f"Median 90-day impressions: {df['impressions_90d'].median():,.0f}")
print(f"Currently trending down: {declining_pages:,} pages ({declining_share:.1%})")
```
