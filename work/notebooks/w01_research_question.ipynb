## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
Question: Which content pages should be prioritized for review based on evidence from search and engagement data?
Decision: Decide which pages should be reviewed first for content refresh or optimization.
Action: The SEO or content team reviews the highest-ranked pages and decides whether to refresh, expand, rewrite, monitor, or leave them unchanged.
Unit of Analysis: One content page.
Output: A ranked list of pages with an opportunity score, suggested action, confidence level, and reason codes.
Cost of a Wrong Recommendation: If the model ranks poor candidates too highly, reviewers waste time on pages that have little improvement potential. If it misses important pages, valuable traffic opportunities may be overlooked.
Why ML Helps: Large websites may contain thousands of pages. Machine learning can combine multiple observable signals, such as impressions, CTR, engagement, freshness, and position, to prioritize pages more consistently than manual review alone.
```
